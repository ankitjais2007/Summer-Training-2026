<h3>2. Build a movie ticket pricing system: children (under 12) pay 50%, students with a valid ID pay 25% less, and Tuesdays have an extra flat 10% off — combine all applicable discounts.</h3>

```python
age=int(input("Enter your age"))
hasStudentId=False;
day='tuesday';
discount_per=0;
ticket_price=500;

if age<12:
    discount_per=50;
    print(" you got  50% off");
elif hasStudentId:
    discount_per=25;
    print("you got extra 25% off");
if day=='tuesday':
    discount_per=discount_per+10;
    print("wow! you got extra 10% off");


total_discount=ticket_price*discount_per/100;
if total_discount==0:
    print(f" pay: {ticket_price}");
else:
    print(f" price: {ticket_price}");
    print(f" total discount: {total_discount}");
    print(f" pay: {ticket_price-total_discount}");
```

<h3>4. Using match-case, build a simple traffic signal simulator that takes "red", "yellow", or "green" and prints the correct instruction, with a default case for invalid input.</h3>

```python
signal=input("Enter signal-color : ");

match signal:
    case 'red': print("stop");
    case 'yellow': print("get ready..");
    case 'green': print("go");
    case _: print("invalid signal");
```
